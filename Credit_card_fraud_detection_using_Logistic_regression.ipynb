```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
creditcartdata = pd.read_csv('creditcard.csv')
```

```python
creditcartdata.head(5)
```

```python
creditcartdata.tail(5)
```

```python
creditcartdata.info()
```

```python
#checking the number of missing values in each column

creditcartdata.isnull().sum()
```

```python
#droping missing values

creditcartdata.dropna(inplace=True)
```

```python
creditcartdata.isnull().sum()
```

```python
#checking the distribution of legit transaction and fraudulent transaction

creditcartdata['Class'].value_counts()
```

seperating the data for analysis

0 ---> Normal transaction
1 ---> fraud transaction

```python
legit_cases = creditcartdata[creditcartdata.Class == 0]
fraud_cases = creditcartdata[creditcartdata.Class  == 1]
```

```python
legit_cases.shape
```

```python
fraud_cases.shape
```

```python
legit_cases['Amount'].describe()
```

```python
fraud_cases['Amount'].describe()
```

```python
#compare the values

creditcartdata.groupby('Class').mean()
```

Under-Sampling

Build a sample dataset containing similar distribution of normal transaction and fradulent transaction

Number of fradulent transaction -----> 104

```python
legist_sample = legit_cases.sample(n=104)
```

Concantenate 2 DF

```python
new_dataset = pd.concat([legist_sample, fraud_cases], axis=0)
```

```python
new_dataset.shape
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

Splitting the data into feature and target

```python
X = new_dataset.drop(columns='Class',axis=1)
Y = new_dataset['Class']
```

```python
print(X)
```

```python
Y
```

Split the data into training and test data

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Training the model

Logistic regression

```python
model = LogisticRegression()
```

```python
#training the logistic regression model

model.fit(X_train,Y_train)
```

Model Evualuation

```python
#Accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
```

```python
training_data_accuracy
```

```python
#Accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
```
